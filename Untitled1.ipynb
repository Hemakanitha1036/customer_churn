```python
# STEP 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# STEP 2: Load Dataset
# Upload customer_churn_sample.csv to your working folder
df = pd.read_csv("customer_churn_sample.csv")

print("DATASET PREVIEW")
print(df.head())

print("\nDataset Shape:", df.shape)
```

```python
# STEP 3: DATA PROFILING
# ============================================

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nDATA TYPES")
print(df.dtypes)

print("\nDATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

print("\nDUPLICATE CUSTOMER IDs")
print(df["CustomerID"].duplicated().sum())

print("\nSUMMARY STATISTICS")
print(df.describe(include="all"))

# Detect outliers using IQR method
numeric_cols = df.select_dtypes(include=np.number).columns

print("\nOUTLIER DETECTION")

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) | (df[col] > upper)
    ]

    print(f"{col}: {len(outliers)} potential outliers")
```

```python
# STEP 4: DATA CLEANING
# ============================================

clean_df = df.copy()

# Remove completely duplicate rows
clean_df = clean_df.drop_duplicates()

# Standardize text columns
text_cols = clean_df.select_dtypes(include="object").columns

for col in text_cols:
    clean_df[col] = clean_df[col].astype("string").str.strip()
    clean_df[col] = clean_df[col].replace("", pd.NA)

# Treat CustomerID as a unique identifier
# Keep the first occurrence of repeated IDs
clean_df = clean_df.drop_duplicates(
    subset=["CustomerID"],
    keep="first"
)

# Fill missing numeric values with median
# Assumption: Median is robust to extreme values
numeric_cols = clean_df.select_dtypes(
    include=np.number
).columns

for col in numeric_cols:
    clean_df[col] = clean_df[col].fillna(
        clean_df[col].median()
    )

# Fill missing categorical values with mode
# Assumption: Most frequent category is representative
categorical_cols = clean_df.select_dtypes(
    include=["object", "string"]
).columns.drop("CustomerID")

for col in categorical_cols:
    mode = clean_df[col].mode(dropna=True)
    if not mode.empty:
        clean_df[col] = clean_df[col].fillna(mode.iloc[0])

# Remove records with missing CustomerID
clean_df = clean_df.dropna(subset=["CustomerID"])

# Flag potential outliers rather than deleting them
# Assumption: Extreme values may represent real customers
for col in numeric_cols:
    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1

    clean_df[col + "_Outlier"] = (
        (clean_df[col] < Q1 - 1.5 * IQR) |
        (clean_df[col] > Q3 + 1.5 * IQR)
    )

print("\nCLEANING COMPLETED")
```

```python
# STEP 5: DATA VALIDATION
# ============================================

assert clean_df["CustomerID"].is_unique, \
    "Duplicate Customer IDs found"

assert clean_df.isnull().sum().sum() == 0, \
    "Missing values remain"

assert (clean_df["Age"] >= 0).all(), \
    "Invalid age found"

assert (clean_df["TenureMonths"] >= 0).all(), \
    "Invalid tenure found"

assert (clean_df["MonthlyCharges"] >= 0).all(), \
    "Invalid monthly charges found"

assert (clean_df["TotalCharges"] >= 0).all(), \
    "Invalid total charges found"

assert (clean_df["SupportTickets"] >= 0).all(), \
    "Invalid support ticket count found"

assert clean_df["Churn"].isin(["Yes", "No"]).all(), \
    "Invalid churn category found"

print("All validation checks passed!")
```

```python
# STEP 6: SUMMARY STATISTICS
# ============================================

print("\nCLEANED DATA SUMMARY")
print(clean_df.describe())

print("\nCHURN DISTRIBUTION")
print(clean_df["Churn"].value_counts())

print("\nCHURN PERCENTAGE")
print(
    clean_df["Churn"].value_counts(normalize=True) * 100
)

print("\nAVERAGE MONTHLY CHARGES BY CHURN")
print(
    clean_df.groupby("Churn")["MonthlyCharges"].mean()
)

print("\nAVERAGE TENURE BY CHURN")
print(
    clean_df.groupby("Churn")["TenureMonths"].mean()
)

print("\nCONTRACT TYPE VS CHURN")
print(
    pd.crosstab(
        clean_df["ContractType"],
        clean_df["Churn"]
    )
)
```

```python
# STEP 7: EXPLORATORY DATA ANALYSIS
# ============================================

# Chart 1: Churn Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=clean_df, x="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Customer Count")
plt.tight_layout()
plt.show()

# Chart 2: Age Distribution
plt.figure(figsize=(7, 4))
sns.histplot(
    data=clean_df,
    x="Age",
    bins=8,
    kde=True
)
plt.title("Customer Age Distribution")
plt.tight_layout()
plt.show()

# Chart 3: Monthly Charges by Churn
plt.figure(figsize=(7, 4))
sns.boxplot(
    data=clean_df,
    x="Churn",
    y="MonthlyCharges"
)
plt.title("Monthly Charges vs Churn")
plt.tight_layout()
plt.show()

# Chart 4: Contract Type vs Churn
plt.figure(figsize=(8, 5))
sns.countplot(
    data=clean_df,
    x="ContractType",
    hue="Churn"
)
plt.title("Contract Type vs Customer Churn")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# Chart 5: Tenure Distribution
plt.figure(figsize=(7, 4))
sns.histplot(
    data=clean_df,
    x="TenureMonths",
    bins=8,
    kde=True
)
plt.title("Customer Tenure Distribution")
plt.tight_layout()
plt.show()

# Chart 6: Correlation Heatmap
plt.figure(figsize=(9, 6))

correlation_cols = [
    "Age",
    "TenureMonths",
    "MonthlyCharges",
    "TotalCharges",
    "SupportTickets"
]

sns.heatmap(
    clean_df[correlation_cols].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Numerical Feature Correlation")
plt.tight_layout()
plt.show()
```

```python
# STEP 8: EXPORT CLEANED DATA
# ============================================

clean_df.to_csv(
    "cleaned_customer_churn.csv",
    index=False
)

print("\nCleaned dataset saved successfully!")
print("Original Rows:", len(df))
print("Cleaned Rows:", len(clean_df))
print("Final Missing Values:",
      clean_df.isnull().sum().sum())
print("Final Duplicate IDs:",
      clean_df["CustomerID"].duplicated().sum())
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import plotly.express as px

# Page configuration
st.set_page_config(
    page_title="Business KPI Dashboard",
    page_icon="📊",
    layout="wide"
)

st.title("📊 Business KPI Dashboard")
st.subheader("Customer Churn and Retention Analysis")

# Load dataset
df = pd.read_csv("customer_churn_sample (1).csv.csv")

# Clean column names
df.columns = df.columns.str.strip()

df["Churn"] = df["Churn"].astype(str).str.strip().str.lower()

# Sidebar filters
st.sidebar.header("Dashboard Filters")

contract = st.sidebar.multiselect(
    "Contract Type",
    df["ContractType"].unique(),
    default=df["ContractType"].unique().tolist()
)

payment = st.sidebar.multiselect(
    "Payment Method",
    df["PaymentMethod"].unique(),
    default=df["PaymentMethod"].unique().tolist()
)

tenure = st.sidebar.slider(
    "Customer Tenure",
    int(df["TenureMonths"].min()),
    int(df["TenureMonths"].max()),
    (
        int(df["TenureMonths"].min()),
        int(df["TenureMonths"].max())
    )
)

# Apply filters
data = df[
    (df["ContractType"].isin(contract)) &
    (df["PaymentMethod"].isin(payment)) &
    (df["TenureMonths"].between(tenure[0], tenure[1]))
].copy()

if data.empty:
    st.warning("No customers found. Change the filters.")
    st.stop()

# KPI calculations
total_customers = len(data)

churned = data[data["Churn"] == "yes"]
retained = data[data["Churn"] == "no"]

churn_rate = len(churned) / total_customers * 100
retention_rate = len(retained) / total_customers * 100

revenue_risk = churned["MonthlyCharges"].sum()

average_tenure = data["TenureMonths"].mean()

# KPI cards
st.subheader("Business Performance")

c1, c2, c3, c4, c5 = st.columns(5)

c1.metric("Total Customers", total_customers)
c2.metric("Churn Rate", f"{churn_rate:.2f}%")
c3.metric("Retention Rate", f"{retention_rate:.2f}%")
c4.metric("Revenue at Risk", f"₹{revenue_risk:,.2f}")
c5.metric("Average Tenure", f"{average_tenure:.1f} Months")

st.divider()

# Churn analysis
st.subheader("1. Customer Churn Analysis")

churn_data = pd.DataFrame({
    "Status": ["Churned", "Retained"],
    "Customers": [len(churned), len(retained)]
})

fig1 = px.pie(
    churn_data,
    names="Status",
    values="Customers",
    title="Churn vs Retention",
    hole=0.4,
    color="Status",
    color_discrete_map={
        "Churned": "#E76F51",
        "Retained": "#2A9D8F"
    }
)

st.plotly_chart(fig1, use_container_width=True)

# Contract analysis
st.subheader("2. Contract Type Analysis")

contract_data = data.groupby("ContractType").apply(
    lambda x: (x["Churn"] == "yes").mean() * 100
).reset_index(name="Churn Rate")

fig2 = px.bar(
    contract_data,
    x="ContractType",
    y="Churn Rate",
    title="Churn Rate by Contract Type",
    text_auto=".1f"
)

st.plotly_chart(fig2, use_container_width=True)

# Revenue analysis
st.subheader("3. Revenue Analysis")

revenue_data = data.groupby("Churn")[
    "MonthlyCharges"
].sum().reset_index()

fig3 = px.bar(
    revenue_data,
    x="Churn",
    y="MonthlyCharges",
    color="Churn",
    title="Monthly Revenue by Customer Status"
)

st.plotly_chart(fig3, use_container_width=True)

# Payment analysis
st.subheader("4. Payment Method Analysis")

payment_data = data.groupby("PaymentMethod")[
    "MonthlyCharges"
].sum().reset_index()

fig4 = px.bar(
    payment_data,
    x="PaymentMethod",
    y="MonthlyCharges",
    title="Revenue by Payment Method",
    text_auto=".2f"
)

st.plotly_chart(fig4, use_container_width=True)

# Customer segmentation
st.subheader("5. Customer Segmentation")

data["Customer Segment"] = pd.cut(
    data["TenureMonths"],
    bins=[-1, 12, 24, float("inf")],
    labels=[
        "New Customers",
        "Growing Customers",
        "Loyal Customers"
    ]
)

segment_data = data.groupby(
    "Customer Segment",
    observed=False
).size().reset_index(name="Customers")

fig5 = px.bar(
    segment_data,
    x="Customer Segment",
    y="Customers",
    color="Customer Segment",
    title="Customer Segments"
)

st.plotly_chart(fig5, use_container_width=True)

# Tenure analysis
st.subheader("6. Tenure Analysis")

fig6 = px.histogram(
    data,
    x="TenureMonths",
    color="Churn",
    title="Customer Distribution by Tenure",
    labels={"TenureMonths": "Tenure (Months)"}
)

st.plotly_chart(fig6, use_container_width=True)

# Customer details
st.subheader("7. Customer Details")

st.dataframe(data, use_container_width=True)

# Download option
st.download_button(
    "Download Customer Data",
    data=data.to_csv(index=False),
    file_name="customer_dashboard_data.csv",
    mime="text/csv"
)

# Business insights
st.subheader("8. Business Insights")

st.write(f"Total Customers: {total_customers}")
st.write(f"Churn Rate: {churn_rate:.2f}%")
st.write(f"Retention Rate: {retention_rate:.2f}%")
st.write(f"Revenue at Risk: ₹{revenue_risk:,.2f}")
st.write(f"Average Tenure: {average_tenure:.2f} months")

st.success("Business KPI Dashboard Completed!")
```

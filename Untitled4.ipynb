```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from matplotlib.backends.backend_pdf import PdfPages
from pathlib import Path

sns.set_theme(style="whitegrid")

# Load your already uploaded dataset
df = pd.read_csv("customer_churn_sample (1).csv.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
# Clean column names
df.columns = df.columns.str.strip()

# Remove duplicate customer records
df = df.drop_duplicates(
    subset="CustomerID"
).copy()

# Convert numeric columns
numeric_columns = [
    "Age",
    "TenureMonths",
    "MonthlyCharges",
    "TotalCharges",
    "SupportTickets"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df.loc[df[col] < 0, col] = np.nan
    df[col] = df[col].fillna(df[col].median())

# Clean churn values
df["Churn"] = (
    df["Churn"].astype(str).str.strip().str.lower()
)

df = df[df["Churn"].isin(["yes", "no"])].copy()

print("Data cleaning completed!")
print(df.isnull().sum())
```

```python
total_customers = len(df)

churned = df[df["Churn"] == "yes"]
retained = df[df["Churn"] == "no"]

churn_count = len(churned)
retained_count = len(retained)

churn_rate = churn_count / total_customers * 100
retention_rate = retained_count / total_customers * 100

revenue_at_risk = churned["MonthlyCharges"].sum()

total_monthly_revenue = df["MonthlyCharges"].sum()

average_tenure = df["TenureMonths"].mean()

print("EXECUTIVE BUSINESS KPIs")
print("-----------------------")
print("Total Customers:", total_customers)
print("Churned Customers:", churn_count)
print("Retained Customers:", retained_count)
print("Churn Rate:", round(churn_rate, 2), "%")
print("Retention Rate:", round(retention_rate, 2), "%")
print("Revenue at Risk: ₹", round(revenue_at_risk, 2))
print("Average Tenure:", round(average_tenure, 2), "Months")
```

```python
# 1. Churn by contract type
contract_analysis = df.groupby("ContractType").agg(
    Customers=("CustomerID", "count"),
    Churned=("Churn", lambda x: (x == "yes").sum())
)

contract_analysis["ChurnRate"] = (
    contract_analysis["Churned"] /
    contract_analysis["Customers"] * 100
)

# 2. Churn by tenure
df["TenureGroup"] = pd.cut(
    df["TenureMonths"],
    bins=[-1, 6, 12, 24, float("inf")],
    labels=["0-6 Months", "7-12 Months",
            "13-24 Months", "Above 24 Months"]
)

tenure_analysis = df.groupby(
    "TenureGroup", observed=False
).agg(
    Customers=("CustomerID", "count"),
    Churned=("Churn", lambda x: (x == "yes").sum())
)

tenure_analysis["ChurnRate"] = (
    tenure_analysis["Churned"] /
    tenure_analysis["Customers"].replace(0, np.nan) * 100
)

# 3. Support ticket analysis
support_analysis = df.groupby("Churn").agg(
    AverageTickets=("SupportTickets", "mean")
)

# 4. Payment method analysis
payment_analysis = df.groupby("PaymentMethod").agg(
    Customers=("CustomerID", "count"),
    Churned=("Churn", lambda x: (x == "yes").sum())
)

payment_analysis["ChurnRate"] = (
    payment_analysis["Churned"] /
    payment_analysis["Customers"] * 100
)

# 5. Revenue risk
revenue_analysis = df.groupby("Churn")[
    "MonthlyCharges"
].sum()

print("INSIGHT 1: Overall churn rate")
print(f"Churn rate: {churn_rate:.2f}%")

print("\nINSIGHT 2: Contract analysis")
display(contract_analysis)

print("\nINSIGHT 3: Tenure analysis")
display(tenure_analysis)

print("\nINSIGHT 4: Support ticket analysis")
display(support_analysis)

print("\nINSIGHT 5: Revenue risk")
print(revenue_analysis)
```

```python
# Revenue recovery scenarios
low_recovery = revenue_at_risk * 0.10
high_recovery = revenue_at_risk * 0.25

print("BUSINESS IMPACT ESTIMATION")
print("--------------------------")

print(
    "Current Monthly Revenue at Risk: ₹",
    round(revenue_at_risk, 2)
)

print(
    "10% Recovery Scenario: ₹",
    round(low_recovery, 2)
)

print(
    "25% Recovery Scenario: ₹",
    round(high_recovery, 2)
)

print("\nEstimated monthly recovery range:")
print(
    f"₹{low_recovery:.2f} to ₹{high_recovery:.2f}"
)
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    data=support_analysis.reset_index(),
    x="Churn",
    y="AverageTickets"
)

plt.title("Average Support Tickets by Churn Status")
plt.xlabel("Customer Status")
plt.ylabel("Average Support Tickets")

plt.show()
```
